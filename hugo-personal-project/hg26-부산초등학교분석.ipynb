## 2026년 부산 초등학교 학년별·학급별 학생 수 불러오기

CSV 파일을 판다스 데이터프레임으로 읽습니다.


```python
import pandas as pd
```

```python
csv_path = '2026년도_학년별·학급별 학생수(초)_부산광역시교육청.csv'
df_students = pd.read_csv(csv_path, encoding='utf-8-sig')

df_students.head()
```

```python
columns_to_keep = [
    "시도교육청",
    "지역",
    "정보공시 학교코드",
    "학교명",
    "설립구분",
    "제외여부",
    "1학년 학생수",
    "2학년 학생수",
    "3학년 학생수",
    "4학년 학생수",
    "5학년 학생수",
    "6학년 학생수",
    "특수학급 학생수",
    "순회학급 학생수",
    "학급수(계)",
    "학생수(계)",
    "교사수",
]

df_students = df_students[columns_to_keep]
```

```python
df_students.info()
```

```python
df_students
```

```python
df_locations = pd.read_csv(
    "전국초중등학교위치표준데이터.csv",
    encoding="utf-8-sig"
)

df_locations.info()
```

```python
df_busan_elementary = df_locations[
    (df_locations["소재지지번주소"].str.contains("부산", na=False))
    & (df_locations["학교급구분"] == "초등학교")
].copy()
```

```python
df_busan_elementary.info()
```

```python
df_merged = pd.merge(
    df_students,
    df_busan_elementary,
    on="학교명",
    how="inner"
)

df_merged.head()
```

```python
df_merged.info()
```

```python
# 각 데이터프레임에서 중복된 학교명 확인
print("students 중복:", df_students["학교명"].duplicated().sum())
print("locations 중복:", df_busan_elementary["학교명"].duplicated().sum())

print(df_students[df_students["학교명"].duplicated(keep=False)]["학교명"])
print(df_busan_elementary[
    df_busan_elementary["학교명"].duplicated(keep=False)
]["학교명"])
```

```python
import folium
import json

from folium import Marker
from folium.plugins import MarkerCluster
```

```python
# 좌표가 없는 행은 제외하고 지도에 표시
schools = df_merged.dropna(subset=["위도", "경도"]).copy()

# 좌표를 숫자로 변환
schools["위도"] = pd.to_numeric(schools["위도"], errors="coerce")
schools["경도"] = pd.to_numeric(schools["경도"], errors="coerce")
schools = schools.dropna(subset=["위도", "경도"])

# 부산 중심 좌표로 지도 생성
school_map = folium.Map(location=[35.1796, 129.0756], zoom_start=11)

# 학교별 마커 추가
for _, row in schools.iterrows():
    folium.Marker(
        location=[row["위도"], row["경도"]],
        tooltip=row["학교명"],
        popup=f'{row["학교명"]}<br>학생 수: {row["학생수(계)"]}명'
    ).add_to(school_map)

school_map
```

```python
# 좌표를 숫자로 바꾸고 유효한 좌표만 남기기
schools = df_merged.copy()
schools["위도"] = pd.to_numeric(schools["위도"], errors="coerce")
schools["경도"] = pd.to_numeric(schools["경도"], errors="coerce")
schools = schools.dropna(subset=["위도", "경도"])

print(f"지도에 표시할 학교 수: {len(schools)}")

# 타일 제공자를 CartoDB로 지정
school_map = folium.Map(
    location=[35.1796, 129.0756],
    zoom_start=11,
    tiles="OpenStreetMap"
)

for _, row in schools.iterrows():
    folium.Marker(
        location=[row["위도"], row["경도"]],
        popup=f'{row["학교명"]}<br>학생 수: {row["학생수(계)"]}명'
    ).add_to(school_map)

school_map
```
